# 1. Strings, indexação e slicing

Uma `str` representa texto.

Considere:

````python
codigo = "PED-2026-0042"
print(codigo)
````

## Indexação

Indexar significa acessar uma posição específica.

````python
codigo = "PED-2026-0042"

print(codigo[0])
print(codigo[1])
print(codigo[-1])
print(codigo[-2])
````

## Slicing

Slicing significa obter uma faixa da sequência.

A forma básica é:

```python
sequencia[inicio:fim]
```

O índice inicial é incluído. O índice final **não** é incluído.

````python
codigo = "PED-2026-0042"

print(codigo[0:3])
print(codigo[4:8])
print(codigo[9:13])
print(codigo[-4:])
````

## Exercício 4 — Processando todos os pedidos

Faça três tarefas:

1. exiba `id`, cliente e produto de cada pedido;
2. calcule o valor total de cada pedido;
3. construa um conjunto com os clientes únicos.

````python
pedidos = [
    {
        "id": "PED-001",
        "cliente": "Ana",
        "produto": "Mouse",
        "quantidade": 2,
        "preco": 120.0
    },
    {
        "id": "PED-002",
        "cliente": "Carlos",
        "produto": "Teclado",
        "quantidade": 1,
        "preco": 250.0
    },
    {
        "id": "PED-003",
        "cliente": "Ana",
        "produto": "Monitor",
        "quantidade": 1,
        "preco": 900.0
    }
]

# Sua solução aqui

print("=== Informações dos pedidos ===")
for pedido in pedidos:
    print(f"ID: {pedido['id']} | Cliente: {pedido['cliente']} | Produto: {pedido['produto']}")

print("\n=== Valores totais por pedido ===")
for pedido in pedidos:
    total = pedido['quantidade'] * pedido['preco']
    print(f"Pedido {pedido['id']}: R$ {total:.2f}")

clientes_unicos = set()
for pedido in pedidos:
    clientes_unicos.add(pedido['cliente'])

print("\n=== Clientes únicos ===")
print(clientes_unicos)
````

# 7. `try` e `except`

Nem toda falha representa um defeito no código.

Considere:

````python
entrada = input("Digite a quantidade: ")

# Teste primeiro com 3.
# Depois teste com "três".

# Remova o #:
# quantidade = int(entrada)
# print(quantidade)
````

`int("3")` funciona.

`int("três")` produz `ValueError`.

Essa é uma situação previsível quando recebemos entrada do usuário.

Podemos tratar:

````python
try:
    quantidade = int(
        input("Digite a quantidade: ")
    )

    print("Quantidade informada:", quantidade)

except ValueError:
    print(
        "Quantidade inválida. Digite um número inteiro."
    )
````

Leia a estrutura:

```python
try:
    # operação que pode gerar uma exceção conhecida

except ValueError:
    # comportamento para esse caso
```

O objetivo não é esconder qualquer erro.

Estamos dizendo:

> Se esta exceção específica acontecer, sei o que fazer.

# 9. Classes, objetos, atributos, métodos e `__init__`

Até agora um pedido poderia ser representado assim:

````python
pedido = {
    "cliente": "Ana",
    "produto": "Mouse",
    "quantidade": 2,
    "preco": 120.0
}

print(pedido)
````

Esse dicionário funciona.

Mas existe também um comportamento relacionado ao pedido:

**calcular seu total**.

Podemos representar dados e comportamento em uma classe.

````python
class Pedido:

    def __init__(
        self,
        cliente,
        produto,
        quantidade,
        preco
    ):
        self.cliente = cliente
        self.produto = produto
        self.quantidade = quantidade
        self.preco = preco

    def calcular_total(self):
        return self.quantidade * self.preco
````

Agora podemos criar um objeto:

````python
pedido = Pedido(
    "Ana",
    "Mouse",
    2,
    120.0
)

print(pedido.cliente)
print(pedido.produto)
print(pedido.preco)
print(pedido.calcular_total())
````

## Vocabulário essencial

```python
class Pedido:
```

### Instanciação

```python
pedido = Pedido(...)
```

É o processo de criação de um objeto a partir da classe.

### Objeto

`pedido` referencia a instância criada.

### Atributo

```python
pedido.preco
```

É um dado associado ao objeto.

### Método

```python
pedido.calcular_total()
```

É um comportamento definido na classe.

### `__init__`

É o método usado aqui para inicializar o estado da instância.

### `self`

Dentro dos métodos, `self` representa a instância sobre a qual o método está operando.
